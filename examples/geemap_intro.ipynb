# Geemap Examples

```python
import ee
import geemap
```

```python
ee.Initialize(project="your-ee-project")
```

```python
m = geemap.Map(center=(20, 0), zoom=2)
dem = ee.Image("USGS/SRTMGL1_003")
countries = ee.FeatureCollection("UN/Geodata/BNDA_simplified/current")
vis_params = {
    "min": 0,
    "max": 4000,
    "palette": ["006633", "E5FFCC", "662A00", "D8D8D8", "F5F5F5"],
}
m.add_layer(dem, vis_params, "SRTM DEM")
style = {"color": "000000ff", "fillColor": "00000000"}
m.add_layer(countries, {}, "Countries")
m
```

```python
collection = ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
```

```python
geometry = ee.Geometry.Point([-122.44297, 37.759859])
images = collection.filterBounds(geometry)
print(images.size().getInfo())
```

```python
images = (
    collection.filterBounds(geometry)
    .filterDate("2025-12-01", "2025-12-31")
    .filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 5))
)
print(images.size().getInfo())
```

```python
m = geemap.Map()
image = images.first()

vis = {
    "min": 0.0,
    "max": 3000,
    "bands": ["B4", "B3", "B2"],
}

m.add_layer(image, vis, "Sentinel-2")
m.center_object(image, 8)
m
```
